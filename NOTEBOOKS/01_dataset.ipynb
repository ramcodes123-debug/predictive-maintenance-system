```python
from pathlib import Path
import sys
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Project root = parent of the notebooks folder
PROJECT_ROOT = Path.cwd()

if PROJECT_ROOT.name == "NOTEBOOKS":
    PROJECT_ROOT = PROJECT_ROOT.parent

DATA_DIR = PROJECT_ROOT / "DATA"
RAW_DIR = DATA_DIR / "RAW"
PROCESSED_DIR = DATA_DIR / "PROCESSED"
MODELS_DIR = PROJECT_ROOT / "MODELS"
OUTPUTS_DIR = PROJECT_ROOT / "OUTPUTS"
PLOTS_DIR = OUTPUTS_DIR / "PLOTS"

# Create only the folders needed for project outputs
for d in [PROCESSED_DIR, MODELS_DIR, PLOTS_DIR]:
    d.mkdir(parents=True, exist_ok=True)

print("Project root:", PROJECT_ROOT)
print("Python:", sys.version)
```

```python
# Search for CSV files
candidates = list(RAW_DIR.rglob("*.csv"))

print("CSV files found:")
for p in candidates[:20]:
    print(" -", p)

if not candidates:
    raise FileNotFoundError(
        "No CSV found. Place the AI4I 2020 predictive maintenance CSV "
        "inside the project's data/raw/ folder."
    )

# Prefer AI4I / predictive-maintenance dataset
preferred = [
    p for p in candidates
    if "ai4i" in p.name.lower()
    or "predictive" in p.name.lower()
]

DATA_PATH = preferred[0] if preferred else candidates[0]

df = pd.read_csv(DATA_PATH)

print("\nUsing:", DATA_PATH)
print("Shape:", df.shape)

display(df.head())
```

```python
print("Columns:")
print(df.columns.tolist())

print("\nData types:")
display(df.dtypes.to_frame("dtype"))

print("\nDataset information:")
df.info()
```

```python
print("Duplicate rows:", df.duplicated().sum())

print("\nTarget distribution:")

target_col = next(
    (
        c for c in df.columns
        if c.lower().replace("_", " ") == "machine failure"
    ),
    None
)

if target_col:
    display(
        df[target_col]
        .value_counts(dropna=False)
        .to_frame("count")
    )
else:
    print("Machine failure target column was not found.")
```
